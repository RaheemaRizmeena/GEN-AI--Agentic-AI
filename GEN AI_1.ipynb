```python
from langchain_ollama import ChatOllama

llm = ChatOllama(
    model="llama3.2",
    temperature=0.6
)

response = llm.invoke(
    "I want to open a clothing store. Suggest some types of dresses."
)

print(response.content)
```

temperature -  It mainly changes how varied/creative the generated text can be/ answers are 
#Low temperature (e.g. 0.1) → more predictable, factual, consistent answers.
#Medium (e.g. 0.6) → a balance between predictable and creative.
#High (e.g., 1.0) → more creative, varied, sometimes less predictable answers.


##Think of .invoke() as:"Hey AI, process this question and give me the answer."
